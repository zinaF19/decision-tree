**1. The typical interview question**

Q: What is your favourite algorithm?
A: Decision trees.
Q: Can you explain it to me in 2 minutes?

<details><summary>Show answer</summary>

A decision tree is a supervised learning algorithm that makes a prediction by asking a sequence of simple questions about the features. We start at the **root node** and, at each node, compare one feature against a threshold (for example "is X2 less than 34.5?"). The answer sends the data point down one of two branches. We keep following branches until we reach a **leaf**, which holds the prediction.

The tree is **grown** from the training data: at every node we look for the split (a feature and a threshold) that best separates the classes, where "best" is measured by a purity criterion such as **Gini impurity**. We apply this rule greedily, splitting again and again until the leaves are pure or a stopping rule is reached.

In short: decision trees work for both classification and regression, need very little data preparation (no scaling required), and are easy to interpret because the path from root to leaf is an explicit set of rules. Their main weakness is a tendency to overfit, which we manage by limiting the tree's growth or by pruning it.

</details>

**2. Explain the tree terminology: node, threshold, leaf, and stump.**

<details><summary>Show answer</summary>

- **Node:** a point in the tree where the data is tested against a condition and split into two branches. The very first node, at the top, is called the **root node**, and nodes that are neither the root nor a leaf are **internal (decision) nodes**.
- **Threshold:** the value a feature is compared against at a node, for example "X2 <= 34.5". Points that satisfy the condition follow one branch, and the rest follow the other.
- **Leaf:** a terminal node with no further splits. It carries the final prediction, which for classification is the majority class of the training samples that landed there.
- **Stump:** the smallest possible tree, a single split with two leaves (a tree of depth 1). Stumps are often used as weak learners inside ensembles such as boosting.

</details>

**3. Explain how to determine the optimal split.**

<details><summary>Show answer</summary>

At each node we search over **every feature** and **every candidate threshold**. The candidate thresholds are usually the midpoints between adjacent values of a feature once it has been sorted.

For each candidate split we:

1. Divide the samples into the two children it would create.
2. Compute the impurity (for example the Gini impurity) of each child.
3. Take the **weighted average** of the two child impurities, weighting by the number of samples in each child.

The **best split** is the one that reduces the impurity the most compared with the parent node. This reduction is called the **information gain**:

```
information gain = impurity(parent) - weighted average impurity(children)
```

We pick the split with the largest gain, then repeat the same procedure greedily at each child node.

</details>

**4. Explain what the Gini impurity is and how to calculate it.**

<details><summary>Show answer</summary>

The **Gini impurity** measures how mixed the classes are within a node. For a node whose class proportions are p_1, p_2, ..., p_k:

```
Gini = 1 - (p_1^2 + p_2^2 + ... + p_k^2)
```

- It equals **0** when the node is **pure** (every sample belongs to the same class).
- It **increases** as the classes become more evenly mixed.
- With more classes the maximum possible value is higher, so a three-class node can be more impure than a two-class one.

**Worked example.** A node holds 9 samples: 4 of class 1, 3 of class 2, and 2 of class 3.

```
Gini = 1 - (4/9)^2 - (3/9)^2 - (2/9)^2 = 0.64
```

</details>

**5. Explain how to make predictions for unseen data (especially with an impure leaf).**

<details><summary>Show answer</summary>

To predict for a new data point, send it down the tree: at each node compare the relevant feature against the threshold and follow the matching branch, until the point reaches a **leaf**.

- **Classification:** the prediction is the **majority class** of the training samples that ended up in that leaf.
- **Impure leaf:** a leaf can contain more than one class (for example when we stop growing early to avoid overfitting). We still predict the majority class. The class proportions in the leaf can additionally be read as **predicted probabilities**, for instance a leaf with 3 samples of class 1 and 1 of class 2 gives a 75% probability for class 1.
- **Regression:** the prediction is the **mean** of the target values of the training samples in the leaf.

</details>

**6. How to avoid overfitting decision trees.**

<details><summary>Show answer</summary>

Overfitting happens when a tree grows so deep that it memorises the noise in the training data instead of the general pattern. We can control this by:

- Limiting the tree depth (`max_depth`).
- Requiring a minimum number of samples before a node may be split (`min_samples_split`) or before a leaf may be formed (`min_samples_leaf`).
- Capping the number of leaves (`max_leaf_nodes`).
- Applying **cost-complexity pruning** (`ccp_alpha`), which trims back branches that add little predictive value.
- Choosing these hyperparameters with **cross-validation** rather than by eye.
- Using an **ensemble** of trees, such as a random forest or gradient boosting, which averages many trees to reduce variance.

</details>

**7. What are the advantages and disadvantages of decision trees?**

<details><summary>Show answer</summary>

| Advantages | Disadvantages |
|---|---|
| Easy to interpret and visualise (a "white-box" model) | Prone to overfitting if growth is left unconstrained |
| No feature scaling or normalisation required | Unstable: a small change in the data can produce a very different tree (high variance) |
| Handle numerical and categorical features | The greedy, step-by-step splitting does not guarantee a globally optimal tree |
| Capture non-linear relationships between features and target | Biased towards features that offer many possible split points |
| Fast to train and fast to predict | Decision boundaries are axis-aligned (step-like), so they fit smooth or diagonal boundaries poorly |
| Need little data preparation | Can struggle with strongly imbalanced classes |

</details>

<details>
<summary>Show answer</summary>

We grow the tree by hand, adding one split at a time and computing the Gini impurity at every node. The guiding principle is straightforward: at each node we choose the split that lowers the impurity the most, and we keep dividing a node only while it remains impure, meaning it still contains more than one class. A node that is already pure needs no further division and becomes a leaf.

### The rule for choosing a split

At every node the algorithm considers all available features (here `X1` and `X2`) and, for each feature, every threshold lying between two neighbouring sorted values. For each candidate it measures how far the impurity falls and keeps the candidate with the largest fall. That reduction is the **Gini gain**:

$$
\text{Gini gain} = G(\text{parent}) - G(\text{weighted children})
$$

The impurity after a split is the average of the two child impurities, weighted by the number of samples that land in each child:

$$
G(\text{weighted children}) = \frac{n_{left}}{n_{total}}\,G(left) + \frac{n_{right}}{n_{total}}\,G(right)
$$

<details>
<summary><strong>Which thresholds are actually tested?</strong></summary>

For each feature we sort the values and place a candidate threshold midway between each pair of neighbours.

**Feature `X1`**

Sorted values:
$$ 15,\ 17,\ 23,\ 24,\ 27,\ 28,\ 33 $$
Candidate thresholds:
$$ 16,\ 20,\ 23.5,\ 25.5,\ 27.5,\ 30.5 $$

**Feature `X2`**

Sorted values:
$$ 33,\ 34,\ 35,\ 37,\ 45 $$
Candidate thresholds:
$$ 33.5,\ 34.5,\ 36,\ 41 $$

The algorithm evaluates the Gini gain for each of these candidates and selects the largest. Working through them, the winner at the root is `X2 ≤ 34.5`, which we examine in Step 1.

</details>

### Step 0: the root node

All nine samples begin in a single node. Their class distribution is:

* Class 1: 4 samples
* Class 2: 3 samples
* Class 3: 2 samples

so the impurity of the root is:

$$
G(\text{parent}) = 1 - \left(\frac49\right)^2 - \left(\frac39\right)^2 - \left(\frac29\right)^2 = 0.64
$$

The value sits above 0.5 because three classes are mixed together.

### Step 1: split on `X2 ≤ 34.5`

Of all the candidates, this threshold gives the largest Gini gain. It isolates the two class-3 points (the only samples with `X2` below 35) in a pure leaf and leaves the remaining seven samples in a node that is still impure:

| Branch | Samples | Classes | Gini |
| ------ | ------: | ------- | ---: |
| `True` (`X2 ≤ 34.5`) | 2 | (3,3) | 0 |
| `False` (`X2 > 34.5`) | 7 | (1,1,1,1,2,2,2) | 0.49 |

$$
G(\text{weighted children}) = \frac29(0) + \frac79(0.49) = 0.38
\qquad\Rightarrow\qquad
\text{gain} = 0.64 - 0.38 = 0.26
$$

The left branch is pure, so it becomes a leaf predicting class 3. The right branch is not, so it is the node we must divide next.

### Step 2: split the impure node on `X2 ≤ 41`

Working only with the seven samples in the right branch, the best available split is `X2 ≤ 41`. It does not yet produce pure children, but it still lowers the impurity, so it is worth making:

| Branch | Samples | Classes | Gini |
| ------ | ------: | ------- | ---: |
| `True` (`X2 ≤ 41`) | 5 | (1,1,1,2,2) | 0.48 |
| `False` (`X2 > 41`) | 2 | (1,2) | 0.50 |

$$
G(\text{weighted children}) = \frac57(0.48) + \frac27(0.50) = 0.485
\qquad\Rightarrow\qquad
\text{gain} = 0.49 - 0.485 = 0.005
$$

The gain is small, and both children remain impure, so each one needs one further split.

### Step 3: split the five-sample child on `X1 ≤ 25.5`

For this node the more informative feature is `X1`. The threshold `X1 ≤ 25.5` separates the three class-1 points (`X1` of 15, 24, 24) from the two class-2 points (`X1` of 27, 33). Both children are now pure, so both become leaves.

### Step 4: split the two-sample child on `X1 ≤ 20`

The same reasoning finishes the other branch. `X1 ≤ 20` sends the single class-1 point (`X1` = 17) one way and the single class-2 point (`X1` = 23) the other, so once again both children are pure.

### The grown tree

![Decision tree grown by hand on the sample data set, showing the split at each node with its class counts and Gini value.](../assets/decision_tree_recap_grown.svg)

Every leaf now has a Gini impurity of 0, which means the tree classifies all nine training samples correctly. With no impure node left, the algorithm stops.

### Could a different tree give the same Gini gain?

Yes. Each threshold is no more than the midpoint between two neighbouring sorted values, so its exact value is not unique: any number lying strictly between the same pair of points produces the identical partition, and therefore the identical Gini gain. The first split, for instance, holds for any threshold between `X2 = 34` and `X2 = 35`, so `X2 ≤ 34.5` and `X2 ≤ 34.7` lead to exactly the same tree. What matters is which points end up together, not the precise threshold value.

</details>

### Evaluate the grown tree with scikit-learn

The final bullet of the exercise asks us to train a decision tree on the same data and plot it, to check the tree we grew by hand. Run the cell below: the tree should reach 100% accuracy on these nine points (the data is perfectly separable) and use the same thresholds we found, `X2 = 34.5`, `X2 = 41`, `X1 = 25.5`, and `X1 = 20`.

You may notice that scikit-learn orders the splits in the right subtree slightly differently from our hand-drawn tree. That is exactly the point raised above: when two splits give the same Gini gain, the greedy algorithm is free to pick either one first, so the decision boundaries are identical even though the drawing differs.

````python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier, plot_tree

# the sample data set from the exercise
data = pd.DataFrame(
    {
        "X1": [15, 17, 24, 24, 23, 27, 33, 27, 28],
        "X2": [35, 45, 37, 37, 45, 37, 37, 34, 33],
        "Y": [1, 1, 1, 1, 2, 2, 2, 3, 3],
    }
)

X = data[["X1", "X2"]]
y = data["Y"]

# grow the tree until every leaf is pure (no depth limit)
tree = DecisionTreeClassifier(random_state=42)
tree.fit(X, y)

print(f"Training accuracy: {tree.score(X, y)}")

# plot the tree as in the earlier notebooks
fig = plt.figure(figsize=(12, 8))
plot_tree(
    tree,
    feature_names=["X1", "X2"],
    class_names=["1", "2", "3"],
    filled=True,
)
plt.show()
````

This is an **extra**: If you want to dive deep into how to build a decision tree **recursively** in Python, check out the following code. 

This code is built like a python script: it starts with the definitions of the classes and their methods. At the very bottom the implemented Decision Tree Classifier is then applied to the iris dataset.




````python
"""Implementation of the CART algorithm to train decision tree classifiers."""
import numpy as np


class Node:
    def __init__(self, predicted_class):
        self.predicted_class = predicted_class
        self.feature_index = 0
        self.threshold = 0
        self.left = None
        self.right = None


class DecisionTreeClassifier:
    def __init__(self, max_depth: int | None = None):
        self.max_depth = max_depth

    def fit(self, X, y):
        """Build decision tree classifier."""
        self.n_classes_ = len(set(y))
        self.n_features_ = X.shape[1]
        self.tree_ = self._grow_tree(X, y)

    def predict(self, X):
        return [self._predict(inputs) for inputs in X]

    def _best_split(self, X, y):
        """Find the best split for a node.
        "Best" means that the average impurity of the two children, weighted by their
        population, is the smallest possible. Additionally it must be less than the
        impurity of the current node.
        To find the best split, we loop through all the features, and consider all the
        midpoints between adjacent training samples as possible thresholds. We compute
        the Gini impurity of the split generated by that particular feature/threshold
        pair, and return the pair with smallest impurity.
        Returns:
            best_idx: Index of the feature for best split, or None if no split is found.
            best_thr: Threshold to use for the split, or None if no split is found.
        """
        # Need at least two elements to split a node.
        m = y.size
        if m <= 1:
            return None, None

        # Count of each class in the current node.
        num_parent = [np.sum(y == c) for c in range(self.n_classes_)]

        # Gini of current node.
        best_gini = 1.0 - sum((n / m) ** 2 for n in num_parent)
        best_idx, best_thr = None, None

        # Loop through all features.
        for idx in range(self.n_features_):
            # Sort data along selected feature.
            thresholds, classes = zip(*sorted(zip(X[:, idx], y)))

            # We could actually split the node according to each feature/threshold pair
            # and count the resulting population for each class in the children, but
            # instead we compute them in an iterative fashion, making this for loop
            # linear rather than quadratic.
            num_left = [0] * self.n_classes_
            num_right = num_parent.copy()
            for i in range(1, m):
                c = classes[i - 1]
                num_left[c] += 1
                num_right[c] -= 1
                gini_left = 1.0 - sum(
                    (num_left[x] / i) ** 2 for x in range(self.n_classes_)
                )
                gini_right = 1.0 - sum(
                    (num_right[x] / (m - i)) ** 2 for x in range(self.n_classes_)
                )

                # The Gini impurity of a split is the weighted average of the Gini
                # impurity of the children.
                gini = (i * gini_left + (m - i) * gini_right) / m

                # The following condition is to make sure we don't try to split two
                # points with identical values for that feature, as it is impossible
                # (both have to end up on the same side of a split).
                if thresholds[i] == thresholds[i - 1]:
                    continue

                if gini < best_gini:
                    best_gini = gini
                    best_idx = idx
                    best_thr = (thresholds[i] + thresholds[i - 1]) / 2

        return best_idx, best_thr

    def _grow_tree(self, X, y, depth: int = 0):
        """Build a decision tree by recursively finding the best split."""
        # Population for each class in current node. The predicted class is the one with
        # largest population.
        num_samples_per_class = [np.sum(y == i) for i in range(self.n_classes_)]
        predicted_class = np.argmax(num_samples_per_class)
        node = Node(predicted_class=predicted_class)

        # Split recursively until maximum depth is reached.
        if self.max_depth is None or depth < self.max_depth:
            idx, thr = self._best_split(X, y)
            if idx is not None:
                indices_left = X[:, idx] < thr
                X_left, y_left = X[indices_left], y[indices_left]
                X_right, y_right = X[~indices_left], y[~indices_left]
                node.feature_index = idx
                node.threshold = thr
                node.left = self._grow_tree(X_left, y_left, depth + 1)
                node.right = self._grow_tree(X_right, y_right, depth + 1)
        return node

    def _predict(self, inputs):
        """Predict class for a single sample."""
        node = self.tree_
        while node.left:
            if inputs[node.feature_index] < node.threshold:
                node = node.left
            else:
                node = node.right
        return node.predicted_class


if __name__ == "__main__":
    from sklearn.datasets import load_iris

    # Load data.
    dataset = load_iris()
    X, y = dataset.data, dataset.target  # pylint: disable=no-member

    # train model
    clf = DecisionTreeClassifier(max_depth=1)
    clf.fit(X, y)

    # predict on one sample
    print(clf.predict([[0, 0, 5, 1.5]]))
````
