```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as pl
```

```python
# create, fit, score, predict.

from sklearn.tree import DecisionTreeClassifier
X = np.array([[2, 2], [2, 1], [2, 3], [1, 2], [1, 1], [3, 3]]) # Features
y = np.array([0, 1, 1, 1, 0, 1]) # Target
tree = DecisionTreeClassifier(random_state=50) # Out of the box.
tree.fit(X, y)
print(f"Node count: {tree.tree_.node_count}, Max depth: {tree.tree_.max_depth}")
# Node count: 9, Max depth: 3
print(tree.score(X, y))
# 1.0
```

```python
from sklearn.tree import plot_tree
plot_tree(tree, filled=True,
feature_names=["Feature 1", "Feature 2"], # instead of x[0], x[1], can use df.columns
class_names=["Class 0", "Class 1"]) # adds the class line
```
